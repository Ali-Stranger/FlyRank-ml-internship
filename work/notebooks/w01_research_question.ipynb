## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

```python
## My Lane: Refresh / Content Opportunity Scoring

I chose the Refresh / Content Opportunity Scoring lane because SEO teams often have limited time and cannot update every page. The goal is to identify which pages should be reviewed first using observable search and content signals such as impressions, CTR, average position, and content freshness.

This is a decision-support project. Instead of refreshing pages randomly, I want to rank pages by their likelihood of benefiting from a review. Machine learning can combine multiple signals better than a single hand-written rule while keeping the recommendations transparent.
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
## Research Question

### Decision
Which pages should be reviewed first for content refresh?

### Action
An SEO manager reviews the ranked list and updates the highest-priority pages first.

### Unit of Analysis
One content page.

### Output
A ranked queue of pages with a priority score and reason codes.

### Cost of a Wrong Recommendation
If a high-potential page is missed, the website may lose search traffic and visibility. If a low-value page is recommended, the team spends time on updates that provide little benefit.

### Why Data and ML Help
The dataset contains multiple observable signals including impressions, clicks, CTR, average position, and content freshness. ML can learn useful combinations of these signals to prioritize pages more effectively than simple rules.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("../data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("Average impressions:", round(df["impressions_90d"].mean(),2))
print("Median days since update:", df["days_since_last_update"].median())
print("Average CTR:", round(df["ctr"].mean(),4))

print("Pages older than 180 days:",
      (df["days_since_last_update"]>=180).sum())

print("Pages with more than 500 impressions:",
      (df["impressions_90d"]>=500).sum())
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
## What I Can and Can't Claim

This project provides decision support by ranking pages that appear to be good candidates for review based on observed search and content signals.

The recommendations do not prove that updating a page will increase rankings or traffic. They identify pages that deserve human review first.

The model cannot predict Google's ranking algorithm. It only learns patterns from the available historical data and should be used alongside expert judgment.
```
